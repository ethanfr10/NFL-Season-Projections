# 00 - NFL Season Projection Model Design

## Core Question

What information is needed to explain why an NFL team wins or loses games, and how can those factors be translated into a reliable season projection?

## Purpose

This notebook defines the football and analytical framework behind the NFL Season Projection Model before predictive modeling begins.

The goal is to make every modeling decision intentional, explainable, and connected to a real football question.

Rather than beginning with available data and searching for useful variables, the project first identifies the major components of team performance and then determines how each component can be measured.

## Design Principles

1. Every feature should represent a meaningful football concept.
2. Team records should be explained through underlying performance rather than treated as the full measure of team quality.
3. Offensive-line and defensive-front performance should be treated as foundational components of every play.
4. Coaching, roster construction, continuity, injuries, schedule, and randomness should be measured separately where possible.
5. Predictions should be understandable to both technical and non-technical audiences.
6. Uncertainty should be reported honestly through probability distributions, simulations, and scenario analysis.
7. The model should be reproducible, updateable, and suitable for historical backtesting.

# Football Knowledge Framework

The model is organized around ten major components of team strength.

Each component represents a different reason a team may improve, decline, win, or lose.

## 1. Offensive Line

The offensive line affects both the passing game and running game.

Planned areas of measurement:

### Pass Protection
- Pressure rate allowed
- Sack rate allowed
- Quarterback hits allowed
- Hurry rate allowed
- Time to pressure
- Pressure allowed without a blitz
- Blitz pickup performance
- Clean-pocket rate
- Pressure attributed to the offensive line where available

### Run Blocking
- Yards before contact
- Stuff rate
- Short-yardage success
- Goal-line success
- Adjusted line yards
- Second-level yards
- Open-field yards
- Run success by direction
- Run success by concept where available

### Continuity and Personnel
- Returning starters
- Returning offensive-line snaps
- Games started together
- New starters
- Rookie starters
- Center and quarterback continuity
- Average age
- Average experience
- Injury history
- Depth and position versatility

## 2. Defensive Front

The defensive front influences quarterback comfort, rushing efficiency, and the amount of pressure placed on the secondary.

Planned areas of measurement:

### Pass Rush
- Pressure rate
- Sack rate
- Quarterback hit rate
- Hurry rate
- Time to pressure
- Four-man pressure rate
- Blitz rate
- Pressure rate when blitzing
- Pressure rate without blitzing

### Run Defense
- Stuff rate
- Rushing EPA allowed
- Rushing success rate allowed
- Yards before contact allowed
- Explosive-run rate allowed
- Short-yardage defense
- Goal-line defense

### Personnel and Rotation
- EDGE quality
- Interior defensive-line quality
- Returning snaps
- Rotation depth
- Average age
- Injury history
- Rookie and free-agent additions

## 3. Passing Game

The passing-game model will evaluate the quarterback, receivers, tight ends, protection, and scheme together.

Planned areas of measurement:

- Passing EPA
- Dropback success rate
- Completion percentage over expected
- Explosive-pass rate
- Air yards
- Yards after catch
- Sack rate
- Interception rate
- Third-down passing
- Red-zone passing
- Play-action performance
- Performance from clean pockets
- Performance under pressure
- Receiver separation or related public measures where available

## 4. Running Game

The running-game model will separate ball-carrier performance from blocking and game situation where possible.

Planned areas of measurement:

- Rushing EPA
- Rushing success rate
- Yards before contact
- Yards after contact
- Explosive-run rate
- Stuff rate
- Short-yardage success
- Goal-line success
- Directional rushing performance
- Designed quarterback rushing
- Scramble production
- Running-back availability and depth

## 5. Secondary and Coverage

The coverage model will evaluate how effectively a defense limits passing efficiency and explosive plays.

Planned areas of measurement:

- Passing EPA allowed
- Dropback success rate allowed
- Completion percentage over expected allowed
- Explosive passes allowed
- Interception rate
- Pass breakup rate where available
- Performance in man and zone coverage where available
- Slot and outside coverage
- Safety support
- Coverage performance with and without pressure
- Returning secondary snaps
- Cornerback and safety depth

## 6. Quarterback Evaluation

Quarterback performance will receive a dedicated projection because of its large effect on team outcomes.

Planned areas of measurement:

- EPA per dropback
- Completion percentage over expected
- Sack avoidance
- Interception rate
- Pressure performance
- Clean-pocket performance
- Deep passing
- Third-down performance
- Red-zone performance
- Scrambling value
- Designed rushing value
- Age
- Experience
- Injury history
- Multi-year performance
- Offensive coordinator continuity
- Offensive-line support
- Receiver support

## 7. Coaching and Organizational Structure

The model will measure coaching and organizational changes rather than assigning subjective labels.

Planned areas of measurement:

- Head coach changes
- Offensive coordinator changes
- Defensive coordinator changes
- New play callers
- Years of play-calling experience
- Previous offensive or defensive efficiency
- Scheme continuity
- Quarterback and coordinator continuity
- Returning starters
- Staff familiarity
- Previous head-coaching experience
- First-time head coaches
- General manager changes
- Major front-office restructuring
- Fourth-down aggressiveness
- Pass rate over expectation
- Pace and situational tendencies

## 8. Roster Construction and Continuity

The roster model will estimate how additions, departures, development, age, and availability change team strength.

Planned areas of measurement:

- Free-agent additions
- Free-agent departures
- Trades
- Retirements
- Draft picks
- Returning injured players
- Returning starters
- Returning snaps
- Returning targets
- Returning carries
- Returning pressures
- Position-group depth
- Average age
- Expected player development
- Injury history
- Projected playing time
- Net roster value added or lost

## 9. Schedule and Matchup Context

The game model will evaluate both overall opponent strength and specific matchup advantages.

Planned areas of measurement:

- Opponent projected strength
- Home-field advantage
- Rest differential
- Short weeks
- Bye weeks
- Consecutive road games
- Travel distance
- International games
- Divisional games
- Weather where appropriate
- Offensive-line versus pass-rush matchup
- Run-blocking versus run-defense matchup
- Receiver versus secondary matchup
- Quarterback versus defensive structure
- Opponent injuries and expected availability

## 10. Randomness and Regression

Some team outcomes are driven by factors that are unlikely to repeat.

Planned areas of measurement:

- One-score record
- Overtime record
- Turnover margin
- Fumble recovery rate
- Defensive touchdowns
- Special-teams touchdowns
- Field-goal variance
- Red-zone overperformance
- Third-down overperformance
- Injury variance
- Expected wins
- Pythagorean wins
- Regression toward league average
